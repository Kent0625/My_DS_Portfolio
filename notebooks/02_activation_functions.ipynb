# Activity 2: Activation Functions & Multi-Layer Perceptron (MLP)

In this activity, we implement and compare core neural network activation functions (Sigmoid, ReLU, and Tanh) and compute a forward pass through a 2-layer feedforward network using NumPy.

```python
import numpy as np

def sigmoid(z):
    return 1.0 / (1.0 + np.exp(-z))

def relu(z):
    return np.maximum(0, z)

def tanh(z):
    return np.tanh(z)

# Sample input vector
z = np.array([-2.0, -0.5, 0.0, 0.5, 2.0])
print("Input:", z)
print("Sigmoid:", np.round(sigmoid(z), 4))
print("ReLU:   ", relu(z))
print("Tanh:   ", np.round(tanh(z), 4))
```

```python
# 2-Layer Feedforward Pass
np.random.seed(42)
X = np.array([[0.5, 1.2]])  # 1 sample, 2 features

# Layer 1: 2 inputs -> 3 hidden neurons
W1 = np.random.randn(2, 3) * 0.1
b1 = np.zeros((1, 3))

# Layer 2: 3 hidden neurons -> 1 output neuron
W2 = np.random.randn(3, 1) * 0.1
b2 = np.zeros((1, 1))

# Forward propagation
H = relu(np.dot(X, W1) + b1)   # Hidden layer with ReLU
y_hat = sigmoid(np.dot(H, W2) + b2)  # Output layer with Sigmoid

print("Hidden Layer Activations:", np.round(H, 4))
print("Final Network Output:    ", np.round(y_hat, 4))
```
