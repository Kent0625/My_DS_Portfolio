$$x = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}, \quad y = \begin{bmatrix} 1 \end{bmatrix}, \quad f(Z_n) = \max(0, Z_n)$$

$$\text{hidden unit weights } W_h = \begin{bmatrix} w_{11}=0.2 & w_{12}=-0.3 \\ w_{13}=0.4 & w_{14}=0.1 \\ w_{15}=-0.5 & w_{16}=0.2 \end{bmatrix}$$

$$\text{output unit weights } W_o = \begin{bmatrix} w_{21}=-0.3 \\ w_{22}=-0.2 \end{bmatrix}, \quad \theta = \begin{bmatrix} \theta_1 = -0.4 \\ \theta_2 = 0.2 \\ \theta_3 = 0.1 \end{bmatrix}$$
</div>

### Step-by-Step Analytical Derivation

#### 1. Hidden Layer Net Inputs ($Z_1, Z_2$)
$$\begin{aligned} Z_1 &= w_{11}x_1 + w_{13}x_2 + w_{15}x_3 + \theta_1 \\     &= (0.2)(1) + (0.4)(0) + (-0.5)(1) + (-0.4) \\     &= 0.2 + 0 - 0.5 - 0.4 = \mathbf{-0.7} \\ Z_2 &= w_{12}x_1 + w_{14}x_2 + w_{16}x_3 + \theta_2 \\     &= (-0.3)(1) + (0.1)(0) + (0.2)(1) + (0.2) \\     &= -0.3 + 0 + 0.2 + 0.2 = \mathbf{0.1} \end{aligned}$$

#### 2. Hidden Layer Activations ($H_1, H_2$)
Applying the ReLU activation function $f(Z) = \max(0, Z)$:
$$\begin{aligned} H_1 &= \max(0, Z_1) = \max(0, -0.7) = \mathbf{0.0} \\ H_2 &= \max(0, Z_2) = \max(0, 0.1) = \mathbf{0.1} \end{aligned}$$

#### 3. Output Layer Net Input ($Z_3$)
$$\begin{aligned} Z_3 &= w_{21}H_1 + w_{22}H_2 + \theta_3 \\     &= (-0.3)(0.0) + (-0.2)(0.1) + 0.1 \\     &= 0.0 - 0.02 + 0.1 = \mathbf{0.08} \end{aligned}$$

#### 4. Predicted Output ($\hat{y}$)
$$\hat{y} = \max(0, Z_3) = \max(0, 0.08) = \mathbf{0.08}$$

#### 5. Error Computations
- **Residual / Output Error ($\delta$):**
  $$\delta = \hat{y} - y = 0.08 - 1.0 = \mathbf{-0.92}$$
- **Squared Error ($E$):**
  $$E = \frac{1}{2}(\hat{y} - y)^2 = \frac{1}{2}(-0.92)^2 = \frac{1}{2}(0.8464) = \mathbf{0.4232}$$
- **Mean Squared Error (MSE):**
  $$\text{MSE} = (\hat{y} - y)^2 = (-0.92)^2 = \mathbf{0.8464}$$

### Code Implementation

````python
import numpy as np

# 1. Define Input and Target from Laboratory Task 2
x = np.array([[1.0, 0.0, 1.0]])  # Shape: (1, 3)
y = np.array([[1.0]])             # Shape: (1, 1)

# 2. Initialize Weights and Biases from Laboratory Task 2
W_h = np.array([
    [ 0.2, -0.3],
    [ 0.4,  0.1],
    [-0.5,  0.2]
])                                # Shape: (3, 2)
b_h = np.array([[-0.4, 0.2]])     # Biases theta1, theta2 (1, 2)

W_o = np.array([
    [-0.3],
    [-0.2]
])                                # Shape: (2, 1)
b_o = np.array([[0.1]])           # Bias theta3 (1, 1)

lr = 0.001                        # Learning rate alpha

# 3. Activation Functions
def relu(z):
    return np.maximum(0, z)

def relu_derivative(z):
    return (z > 0).astype(float)

# ----------------- FORWARD PROPAGATION -----------------
Z_h = np.dot(x, W_h) + b_h
H = relu(Z_h)

Z_o = np.dot(H, W_o) + b_o
y_hat = relu(Z_o)

loss = 0.5 * np.sum((y_hat - y) ** 2)

print("=== FORWARD PROPAGATION RESULTS ===")
print(f"Z_h (Hidden Pre-activation): {Z_h}")
print(f"H   (Hidden Activation):     {H}")
print(f"Z_o (Output Pre-activation): {Z_o}")
print(f"y_hat (Model Prediction):    {y_hat}")
print(f"Initial Loss:                {loss:.6f}\n")

# ----------------- BACKWARD PROPAGATION -----------------
delta_o = (y_hat - y) * relu_derivative(Z_o)
delta_h = np.dot(delta_o, W_o.T) * relu_derivative(Z_h)

dW_o = np.dot(H.T, delta_o)
db_o = delta_o

dW_h = np.dot(x.T, delta_h)
db_h = delta_h

# Gradient Descent Updates
W_o_updated = W_o - lr * dW_o
b_o_updated = b_o - lr * db_o
W_h_updated = W_h - lr * dW_h
b_h_updated = b_h - lr * db_h

print("=== BACKWARD PROPAGATION GRADIENTS ===")
print("dW_o:\n", dW_o)
print("db_o:\n", db_o)
print("dW_h:\n", dW_h)
print("db_h:\n", db_h)

print("\n=== UPDATED PARAMETERS (lr = 0.001) ===")
print("Updated W_o:\n", W_o_updated)
print("Updated b_o:\n", b_o_updated)
print("Updated W_h:\n", W_h_updated)
print("Updated b_h:\n", b_h_updated)
````

    <li>Criterion: MSE Loss</li>
    <li>Fully Connected Layers x 2</li>

    <li>Epoch: 1000</li>

### Code Implementation

````python
import torch
import torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader

# Set seed for reproducible results
torch.manual_seed(42)

# 1. Generate Synthetic Regression Dataset (128 samples, 3 input features, 1 target)
X = torch.randn(128, 3)
y = (2.0 * X[:, 0] - 3.0 * X[:, 1] + 1.5 * X[:, 2] + 0.5).unsqueeze(1) + 0.05 * torch.randn(128, 1)

# 2. Setup DataLoader with specified Batch Size = 8
batch_size = 8
dataset = TensorDataset(X, y)
train_loader = DataLoader(dataset, batch_size=batch_size, shuffle=True)

# 3. Construct Model with 2 Fully Connected Layers
class TwoLayerRegression(nn.Module):
    def __init__(self, in_features=3, hidden_features=16, out_features=1):
        super(TwoLayerRegression, self).__init__()
        self.fc1 = nn.Linear(in_features, hidden_features)   # Layer 1
        self.fc2 = nn.Linear(hidden_features, out_features)  # Layer 2
        
    def forward(self, x):
        out = self.fc1(x)
        out = self.fc2(out)
        return out

model = TwoLayerRegression(in_features=3, hidden_features=16, out_features=1)

# 4. Set Loss Criterion (MSE Loss) and Optimizer (SGD)
criterion = nn.MSELoss()
optimizer = torch.optim.SGD(model.parameters(), lr=0.01)

# 5. Train for 1000 Epochs
epochs = 1000

print("=== BEGINNING TRAINING (1000 EPOCHS, BATCH SIZE 8) ===")
for epoch in range(epochs):
    epoch_loss = 0.0
    for batch_x, batch_y in train_loader:
        # Forward pass
        preds = model(batch_x)
        loss = criterion(preds, batch_y)
        
        # Backward pass & optimization
        optimizer.zero_grad()
        loss.backward()
        optimizer.step()
        
        epoch_loss += loss.item() * batch_x.size(0)
        
    epoch_loss /= len(dataset)
    
    if (epoch + 1) % 100 == 0:
        print(f"Epoch [{epoch + 1:4d}/{epochs}], Loss: {epoch_loss:.6f}")

print(f"\nTraining Complete! Final Epoch Loss: {epoch_loss:.6f}")
````

### Code Implementation

````python
# 1. Perform Standard Imports
import torch
import numpy as np

# 2. Create set_seed() function
def set_seed(seed: int):
    np.random.seed(seed)
    torch.manual_seed(seed)

# 3. Create a NumPy array 'arr' of 6 random integers between 0 and 5
set_seed(42)
arr = np.random.randint(0, 5, 6)
print("3. NumPy array 'arr':", arr)

# 4. Create tensor x from array
x = torch.from_numpy(arr)
print("4. Tensor x:", x)

# 5. Change dtype of x from int32 to int64
print("5. Old Type:", x.type())
x = x.type(torch.int64)
print("   New Type:", x.type())

# 6. Reshape x into a 3x2 tensor
x = x.reshape(3, 2)
print("6. Reshaped x (3x2):\n", x)

# 7. Return the right-hand column of tensor x
print("7. Right-hand column:\n", x[:, 1:])

# 8. Without changing x, return a tensor of square values of x
x_squared = x.pow(2)
print("8. Squared values:\n", x_squared)
print("   Original x unchanged:\n", x)

# 9. Create tensor y with shape (2, 3) to permit matrix multiplication with x (3, 2)
set_seed(42)
y = torch.randint(0, 5, (2, 3))
print("9. Tensor y (2x3):\n", y)

# 10. Find the matrix product of x and y.
prod = torch.mm(x, y)
print("10. Matrix product prod (3x3):\n", prod)
````
