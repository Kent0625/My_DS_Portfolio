### Solution & Inquiries

**Scenario Context:** Dr. Jake is a medical doctor and the CEO of Astro JR, a pharmaceutical company in the Philippines producing and deploying the COVID-19 vaccine worldwide.

#### 1. Descriptive Analytics
*Focus: Summarizing historical records to capture what happened.*
- **Inquiry 1:** What was the total number of AstroJR vaccine doses distributed and administered across each administrative region of the Philippines over the past quarter?
- **Inquiry 2:** What percentage of vaccinated individuals reported mild or moderate adverse reactions within 48 hours of vaccination across clinical batches?

#### 2. Diagnostic Analytics
*Focus: Investigating underlying root causes and correlations to explain why events occurred.*
- **Inquiry 1:** Why did vaccine uptake and demand drop significantly in specific rural provinces during the second quarter?
- **Inquiry 2:** Which supply chain disruptions (e.g., cold-chain temperature deviations, transit delays) were the primary causes of vaccine dose spoilage during distribution?

#### 3. Predictive Analytics
*Focus: Leveraging statistical and machine learning models to project future outcomes.*
- **Inquiry 1:** What will be the projected global demand for AstroJR booster doses over the next six months based on emerging variant transmission rates?
- **Inquiry 2:** Which geographic areas have the highest probability of experiencing a COVID-19 case surge in the upcoming month?

#### 4. Prescriptive Analytics
*Focus: Recommending optimal courses of action and resource allocations given operational constraints.*
- **Inquiry 1:** How should AstroJR reallocate its cold-storage transport fleet across distribution hubs to minimize delivery times and eliminate vaccine wastage during regional spikes?
- **Inquiry 2:** What is the optimal production schedule and raw material procurement strategy needed to meet anticipated demand surges without exceeding warehouse holding limits?
