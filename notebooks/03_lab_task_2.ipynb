$$x = \begin{bmatrix} 1 \\ 0 \\ 1 \end{bmatrix}, \quad y = \begin{bmatrix} 1 \end{bmatrix}, \quad f(Z_n) = \max(0, Z_n)$$

$$\text{hidden unit weights } W_h = \begin{bmatrix} w_{11}=0.2 & w_{12}=-0.3 \\ w_{13}=0.4 & w_{14}=0.1 \\ w_{15}=-0.5 & w_{16}=0.2 \end{bmatrix}$$

$$\text{output unit weights } W_o = \begin{bmatrix} w_{21}=-0.3 \\ w_{22}=-0.2 \end{bmatrix}, \quad \theta = \begin{bmatrix} \theta_1 = -0.4 \\ \theta_2 = 0.2 \\ \theta_3 = 0.1 \end{bmatrix}$$

### Step-by-Step Analytical Derivation

#### 1. Hidden Layer Net Inputs ($Z_1, Z_2$)
$$\begin{aligned} Z_1 &= w_{11}x_1 + w_{13}x_2 + w_{15}x_3 + \theta_1 \\     &= (0.2)(1) + (0.4)(0) + (-0.5)(1) + (-0.4) \\     &= 0.2 + 0 - 0.5 - 0.4 = \mathbf{-0.7} \\ Z_2 &= w_{12}x_1 + w_{14}x_2 + w_{16}x_3 + \theta_2 \\     &= (-0.3)(1) + (0.1)(0) + (0.2)(1) + (0.2) \\     &= -0.3 + 0 + 0.2 + 0.2 = \mathbf{0.1} \end{aligned}$$

#### 2. Hidden Layer Activations ($H_1, H_2$)
Applying the ReLU activation function $f(Z) = \max(0, Z)$:
$$\begin{aligned} H_1 &= \max(0, Z_1) = \max(0, -0.7) = \mathbf{0.0} \\ H_2 &= \max(0, Z_2) = \max(0, 0.1) = \mathbf{0.1} \end{aligned}$$

#### 3. Output Layer Net Input ($Z_3$)
$$\begin{aligned} Z_3 &= w_{21}H_1 + w_{22}H_2 + \theta_3 \\     &= (-0.3)(0.0) + (-0.2)(0.1) + 0.1 \\     &= 0.0 - 0.02 + 0.1 = \mathbf{0.08} \end{aligned}$$

#### 4. Predicted Output ($\hat{y}$)
$$\hat{y} = \max(0, Z_3) = \max(0, 0.08) = \mathbf{0.08}$$

#### 5. Error Computations
- **Residual / Output Error ($\delta$):**
  $$\delta = \hat{y} - y = 0.08 - 1.0 = \mathbf{-0.92}$$
- **Squared Error ($E$):**
  $$E = \frac{1}{2}(\hat{y} - y)^2 = \frac{1}{2}(-0.92)^2 = \frac{1}{2}(0.8464) = \mathbf{0.4232}$$
- **Mean Squared Error (MSE):**
  $$\text{MSE} = (\hat{y} - y)^2 = (-0.92)^2 = \mathbf{0.8464}$$
